There are 5 main steps that need to be completed for step 2: Create a list, iterate over the list, sort results, generate random numbers, add to the list, and print the results. 

The following creates a list of numbers that I chose. 

```python
numbers = [10, 1, 15, 5, 7, 23, 2, 6]
print("Number list is:", numbers)
```

The main idea of iteration is that the step are done over and over again until the list is complete. In this case, there are a total of 8 numbers, the code will run until the numbers go through the loop.

```python
print("Numbers in the list:")

for number in numbers:
    print(number)
```

Numbers can then be sorted, for this example the numbers can be sorted from smallest to biggest.

```python
numbers.sort() # reverse=true if descending order

print(numbers)
```

Loop can also be used to create additional numbers for the list. 

```python
for i in range(5):
    numbers.append(random.randint(1,1000))
```

```python
numbers.sort()

print("Final Number list:", numbers)
```

The first step is to generate a list of 50 random values. These values will be between 0 and 100. 

```python
random_list = [random.randint(0, 100) for x in range (0,50)]
# randint is saying the generated values must be intergers, values can be between 0-100
# for statement means that it will be done 50

random_list
```

Now that 50 values were generated, these will be used to make a new list, with values that are divisible by three. 

```python
list_by_three = [a for a in random_list if a % 3 == 0]

list_by_three
```

Out of the 50 original values, there are only 17 values that are divisible by three. 

The next step is to find the difference in length between the two list. 

```python
length_random = len(random_list)

length_three = len(list_by_three)

print(length_random)
print(length_three)
```

```python
difference = length_random - length_three

difference
```
