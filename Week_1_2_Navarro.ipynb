Out of the 50 original values, there are only 21 values that are divisible by three. 

The next step is to find the difference in length between the two list. 

```python
length_random = len(random_list)

length_three = len(list_by_three)

print(length_random)
print(length_three)
```

```python
difference = length_random - length_three

difference
```

The first list has a total of 50 values, the second has 21. The difference can be found by subtracting these values from each other. There is a difference of 29 values between the two list. 

The previous steps will now be done 10 times. The output should be a total of 10 different values, which show the difference between the two list. 

```python
number_of_times = 10
difference_list = []

# loop
for i in range(0, number_of_times):
    random_list = [random.randint(0, 100) for x in range (0, 50)]
    list_by_three = [a for a in random_list if a % 3 == 0]

    # start of difference calculation
    length_random = len(random_list)
    length_three = len(list_by_three)
    difference = length_random - length_three
    difference_list.append(difference)

difference_list
```

Now that the differences are calculated, average will be calculated. 

```python
average_diff = sum(difference_list)/ float(len(difference_list))

average_diff
```

One objective for this activity is to count the number of times words appear. Before that can be done,  lines, unwanted spaces, special characters, and punctuation need to be removed from the text

```python
# removal of lines and spacing
text = text.replace('\n', "")
```

```python
# removal of punctuation
cleaned = ""
for char in text:
    if char == " ":
        cleaned += char
    elif char.isalnum():
        cleaned += char
    else:
        cleaned += " "
```

With the cleaned text, a list of unique words can be made, with the count. 

```python
list_of_words = cleaned.split()
list_of_words
```

There is a total of 871 unique words in the first chapter. 

```python
unique_words = dict.fromkeys(list_of_words)
len(list(unique_words.keys()))
```

Each word will be counted to see how many times it shows up in the chapter. This will be done using a for loop. If the word shows up more than onece it will be counted/added. 

```python
for word in list_of_words:
    if unique_words[word] is None:
        unique_words[word] = 1
    else: 
        unique_words[word] += 1
unique_words
```

In this chapter there are many words that only show up once. With that in mind, the top 25 words will be found. These are the words that show up the most. 

```python
top = sorted(unique_words.items(), key=lambda key_val_tuple: key_val_tuple[1], reverse=True)
top[:25]
```

```python
def convert_to_number(number_stack):
    final_number = 0
    for i in range(0, len(number_stack)):
        final_number += (number_stack.pop() * (math.pow(10, i)))
    return final_number

for number_tuple in permutations(range(3)):
    number_stack = list(dropwhile(lambda x: x <= 0, number_tuple))
    print(convert_to_number(number_stack))
```
