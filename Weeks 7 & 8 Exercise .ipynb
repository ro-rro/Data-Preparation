## Ch 8: 

The first step in this chapter is to convert the dataset from wide to a long format. The format will only show the candy that is being rated, the candy, and the respondent number (which should be unique). 

```python
q6 = candy1.filter(regex=r"^Q6\s*\|").columns

candy_long = candy1[q6].reset_index().melt(
    id_vars="index", 
    var_name="Candy", 
    value_name="Rating")

candy_long = candy_long.rename(columns={"index":"Respondent"})
```

```python
candy_long.head()
```

The following only shows the Candy rating for 100 Grand Bar for the first 15 respondents. This is also known as hierachical indexing, it has two levels candy and rating. 

```python
candy_in = candy_long.set_index(["Candy", "Rating"])

candy_in.head(15)
```

Any type of candy and its rating can be found, for example the twix ratings:

```python
candy_in.loc["Q6 | Twix"]
```

The next part of chapter 8 is to create a pivot table. 

```python
candy_pivot = candy_long.pivot_table(
    index="Candy", 
    columns="Rating",
    aggfunc="size", 
    fill_value=0)

candy_pivot.head(20)
```

The information from the pivot table is easier to understand. It has a list of candy and the number of times it was rated despair, meh, or joy. 

## Ch. 10: 
This chapter will focus on grouping with index levels, and doing split, apply, combine. 

The first part we will group by index level. The idea is to see how many candies were ranked dispair, joy, or meh. The first step is to make sure that there are two levels, rating and candy. After that is done it can be grouped by index level (rating). 


```python
candy_summary = candy1[q6].apply(
    lambda col: col.value_counts()).fillna(0).astype(int)
```

```python
candy_summary = candy_summary.stack()

candy_summary.index.names=["Candy", "Rating"]

candy_total = candy_summary.groupby(level="Candy").sum()
candy_total
```

Most are in the joy category (64206) and meh has the least (53095). 

The next thing that will be done in chapter 10 is split, apply, combine. The goal is to see how many times each candy has joy, meh, or despair. In other words, this is the same thing that we have done previously, but it SPLITs the candy by columns, APPLYs the percentage of joy, despair, and meh for each candy, and COMBINES the results in a dataframe.  

```python
rating = candy1[q6].apply(
    lambda col: col.value_counts(normalize=True) * 100)
# multiplied by 100 for percentage

rating
```

All of the candies have different results. There are some that were rated more joy then meh, while other were more despair thant the other options. 

## Ch. 11:
For the last part of the assignment, this chapter will focus on dates. Since this date does not contain any type of dates, the first step would be to generate a date range. 

```python
dates = pd.date_range(
    start="2016-01-25", 
    periods=len(candy1), 
    freq="D")
dates
```

Once all of the dates have been generated they can be added to the dataset. 

```python
candy1["Date_entry"] = dates
candy1.head(10)
```

Now that there is a date for each respondent in the dataset, the dates can now be converted to string, same formate but it goes by month, day, then year. 

```python
candy1["Date_String"] = candy1["Date_entry"].dt.strftime("%m-%d-%Y")

candy1.head(10)
```

Now convert it back to datetime. 

```python
candy1["Date Converted"] = pd.to_datetime(
    candy1["Date_String"])

candy1.head()
```

The dataset now contains all the the date related columns that way the changes can easily be seen.
